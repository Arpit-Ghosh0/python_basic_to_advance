# 3. Functions

Functions are reusable blocks of code. Python functions are **first-class objects** — they
can be assigned to variables, passed as arguments, and returned from other functions.

```python
# =========================================================
# BASIC FUNCTIONS
# =========================================================

def greet(name):
    """Docstring: explains what the function does. Shows up via help(greet)."""
    return f"Hello, {name}!"

print(greet("Alice"))
print(help(greet))   # prints the docstring + signature
```

```python
# =========================================================
# DEFAULT ARGUMENTS, *args, **kwargs
# =========================================================

def power(base, exponent=2):     # exponent has a default value
    return base ** exponent

print(power(4))         # uses default exponent=2 -> 16
print(power(4, 3))      # overrides default        -> 64
print(power(base=4, exponent=3))   # keyword argument call

def total(*args):                 # *args -> collects extra positional args into a tuple
    return sum(args)

print(total(1, 2, 3, 4))   # 10

def show_info(**kwargs):          # **kwargs -> collects extra keyword args into a dict
    for key, value in kwargs.items():
        print(f"{key}: {value}")

show_info(name="Alice", age=30)

def full_demo(a, b, *args, c=10, **kwargs):
    # a, b   -> required positional
    # *args  -> any extra positional args
    # c      -> keyword-only argument with default
    # kwargs -> any extra keyword args
    print(a, b, args, c, kwargs)

full_demo(1, 2, 3, 4, c=99, extra="hi")
```

```python
# =========================================================
# LAMBDA (anonymous) FUNCTIONS
# =========================================================
# Small, single-expression, unnamed functions. Good for short throwaway logic.

square = lambda x: x ** 2
print(square(5))   # 25

add = lambda x, y: x + y
print(add(3, 4))   # 7

# Commonly used inline, e.g. as a sort key
people = [("Alice", 30), ("Bob", 25), ("Eve", 35)]
people.sort(key=lambda person: person[1])   # sort by age
print(people)
```

```python
# =========================================================
# RECURSION - a function that calls itself
# =========================================================

def factorial(n):
    if n <= 1:            # base case - stops the recursion
        return 1
    return n * factorial(n - 1)   # recursive case

print(factorial(5))   # 120

def fibonacci(n, memo={}):
    # memoization (caching results) avoids recomputation - crucial for recursive efficiency
    if n in memo:
        return memo[n]
    if n <= 1:
        return n
    memo[n] = fibonacci(n - 1, memo) + fibonacci(n - 2, memo)
    return memo[n]

print([fibonacci(i) for i in range(10)])
```

```python
# =========================================================
# SCOPE - where a variable is visible/accessible (LEGB rule)
# Local -> Enclosing -> Global -> Built-in
# =========================================================

x = "global"

def outer():
    x = "enclosing"

    def inner():
        x = "local"
        print("inner:", x)

    inner()
    print("outer:", x)

outer()
print("global:", x)

def modify_global():
    global x          # explicitly says: use the GLOBAL x, don't create a local one
    x = "modified global"

modify_global()
print(x)
```
