# 1. Basics

Python is a **dynamically typed**, **interpreted** language. "Dynamically typed" means you
don't declare a variable's type — Python figures it out at runtime based on the value.


```python
# =========================================================
# COMMENTS
# =========================================================

# This is a single-line comment. The interpreter ignores everything after '#'.

"""
This is technically a string literal, not a "real" comment,
but when it's not assigned to anything (and especially right below
a function/class definition) it's used as a MULTI-LINE COMMENT
or a DOCSTRING (documentation string).
"""

print("Hello, World!")  # inline comment explaining the line above
```

```python
# =========================================================
# VARIABLES
# =========================================================
# You don't declare a type - Python infers it from the value you assign.
# Naming convention (PEP8): snake_case for variables and functions.

name = "Alice"       # str
age = 30              # int
height = 5.6           # float
is_student = False    # bool

# Multiple assignment in one line
x, y, z = 1, 2, 3

# Assign the same value to multiple variables
a = b = c = 0

# Swapping variables - no temp variable needed in Python!
x, y = y, x
print("Swapped:", x, y)

# You can always check a variable's type at runtime
print(type(name), type(age), type(height), type(is_student))
```

```python
# =========================================================
# CORE DATA TYPES
# =========================================================

integer_num   = 42                # int      - whole numbers, arbitrary precision
float_num     = 3.14159           # float    - decimal numbers
complex_num   = 2 + 3j             # complex  - has a real and imaginary part
string_val    = "Hello"           # str      - immutable sequence of characters
boolean_val   = True              # bool     - True or False (subclass of int: True == 1)
none_val      = None              # NoneType - represents "no value" / absence of a value

list_val   = [1, 2, 3]             # list  - ordered, MUTABLE, allows duplicates
tuple_val  = (1, 2, 3)             # tuple - ordered, IMMUTABLE, allows duplicates
set_val    = {1, 2, 3}             # set   - unordered, MUTABLE, no duplicates
dict_val   = {"key": "value"}      # dict  - key-value pairs (hash map)

for value in (integer_num, float_num, complex_num, string_val,
              boolean_val, none_val, list_val, tuple_val, set_val, dict_val):
    print(f"{str(value):<20} -> {type(value).__name__}")
```

```python
# =========================================================
# TYPE CASTING (converting between types)
# =========================================================

print(int("42"))        # str -> int
print(float("3.14"))    # str -> float
print(str(100))         # int -> str
print(bool(0), bool(1)) # 0/empty values are Falsy, everything else is Truthy
print(list("abc"))      # str -> list of characters
print(tuple([1, 2, 3])) # list -> tuple

# Falsy values in Python: False, None, 0, 0.0, "", [], (), {}, set()
# Everything else is Truthy
print(bool(""), bool("non-empty"), bool([]), bool([1]))
```

```python
# =========================================================
# OPERATORS
# =========================================================

# Arithmetic
print(7 + 3, 7 - 3, 7 * 3, 7 / 3)   # + - * /  (division ALWAYS returns float)
print(7 // 3)   # floor (integer) division -> 2
print(7 % 3)    # modulo (remainder)       -> 1
print(7 ** 3)   # exponentiation           -> 343

# Comparison -> always return a bool
print(5 == 5, 5 != 4, 5 > 4, 5 < 4, 5 >= 5, 5 <= 4)

# Logical
print(True and False, True or False, not True)

# Identity vs equality - IMPORTANT distinction!
a = [1, 2, 3]
b = [1, 2, 3]
c = a
print(a == b)   # True  -> same VALUES
print(a is b)   # False -> different OBJECTS in memory
print(a is c)   # True  -> c points to the SAME object as a

# Membership
print(2 in [1, 2, 3])       # True
print("x" not in "abc")     # True

# Augmented assignment
n = 10
n += 5   # same as n = n + 5
n -= 2
n *= 2
print(n)
```

```python
# =========================================================
# STRINGS - one of the most-used types, worth its own section
# =========================================================

s = "Python Programming"

print(s.upper())          # PYTHON PROGRAMMING
print(s.lower())          # python programming
print(s.title())          # Python Programming
print(s.replace("Python", "Java"))
print(s.split(" "))       # ['Python', 'Programming'] -> splits into a list
print(" , ".join(["a", "b", "c"]))  # 'a , b , c' -> opposite of split
print(s.strip())          # removes leading/trailing whitespace
print(len(s))             # length of the string
print(s[0], s[-1])        # indexing: first char, last char
print(s[0:6])             # slicing: "Python" (start:stop, stop excluded)
print(s[::-1])            # slicing trick: reverses the string

# f-strings (formatted string literals) - the modern, preferred way to format strings
name, score = "Bob", 95.5
print(f"{name} scored {score:.1f}%")   # :.1f -> 1 decimal place
print(f"{name!r}")                     # !r -> calls repr() -> \'Bob\'
print(f"{2 + 2 = }")                   # self-documenting expressions (Python 3.8+)

# .format() and % formatting also exist (older styles, good to recognize)
print("{} is {} years old".format("Alice", 30))
print("%s is %d years old" % ("Alice", 30))
```

```python
# =========================================================
# INPUT / OUTPUT
# =========================================================

# input() always returns a str - cast it if you need a number
# user_input = input("Enter your name: ")

print("A", "B", "C")                       # default separator is a space
print("A", "B", "C", sep="-")              # custom separator -> A-B-C
print("No newline at the end", end=" ")    # custom end (default is '\n')
print("continues here")
```
