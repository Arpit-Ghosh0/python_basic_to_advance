# 4. Data Structures

Python's built-in containers. Choosing the right one matters a lot for both correctness and
performance.

| Type  | Ordered | Mutable | Duplicates | Typical use |
|-------|---------|---------|------------|-------------|
| list  | yes     | yes     | yes        | general-purpose sequence |
| tuple | yes     | no      | yes        | fixed collection / dict keys |
| set   | no      | yes     | no         | uniqueness, fast membership tests |
| dict  | yes*    | yes     | keys: no   | key -> value lookups |

*dicts preserve insertion order since Python 3.7

```python
# =========================================================
# LISTS
# =========================================================

nums = [5, 3, 8, 1]

nums.append(9)        # add to the end
nums.insert(1, 100)   # insert at a specific index
nums.remove(3)        # remove the FIRST occurrence of a value
popped = nums.pop()   # remove & return the LAST item (or pop(index) for a specific one)
nums.sort()           # sort in-place (ascending)
nums.reverse()        # reverse in-place
print(nums, "popped:", popped)

# Slicing: list[start:stop:step]
letters = ["a", "b", "c", "d", "e"]
print(letters[1:3])    # ['b', 'c']
print(letters[:3])     # ['a', 'b', 'c']
print(letters[::2])    # every 2nd item -> ['a', 'c', 'e']
print(letters[::-1])   # reversed

# Nested lists (e.g. a simple matrix)
matrix = [[1, 2], [3, 4]]
print(matrix[1][0])   # 3

# Shallow copy pitfall - very common bug!
a = [1, 2, 3]
b = a              # b is the SAME list, not a copy!
b.append(4)
print(a)           # a is also changed: [1, 2, 3, 4]

c = a.copy()       # or a[:] or list(a) -> an actual independent copy
c.append(5)
print(a, c)
```

```python
# =========================================================
# TUPLES - immutable, so faster and safer for fixed data
# =========================================================

point = (3, 4)
x, y = point           # unpacking
print(x, y)

single = (5,)           # note the trailing comma - required for a 1-element tuple!
not_a_tuple = (5)       # this is just an int in parentheses

# Tuples are commonly used as dict keys (lists CANNOT be, since they're mutable/unhashable)
locations = {(0, 0): "origin", (1, 1): "diagonal"}
print(locations[(0, 0)])
```

```python
# =========================================================
# SETS - unordered, unique elements, fast membership tests
# =========================================================

s1 = {1, 2, 3, 4}
s2 = {3, 4, 5, 6}

print(s1 | s2)   # union            -> {1,2,3,4,5,6}
print(s1 & s2)   # intersection     -> {3,4}
print(s1 - s2)   # difference       -> {1,2}
print(s1 ^ s2)   # symmetric diff   -> {1,2,5,6} (in one but not both)

s1.add(10)
s1.discard(1)    # like .remove() but doesn't error if the item is missing

# Very common real use: de-duplicating a list while checking membership fast
duplicates = [1, 2, 2, 3, 3, 3]
print(set(duplicates))   # {1, 2, 3}
```

```python
# =========================================================
# DICTIONARIES - key/value pairs, extremely common in real code
# =========================================================

person = {"name": "Alice", "age": 30, "city": "NYC"}

print(person["name"])          # direct access (raises KeyError if missing)
print(person.get("email"))     # safe access -> None if missing (no error)
print(person.get("email", "N/A"))   # safe access with a fallback default

person["email"] = "alice@example.com"   # add/update a key
del person["city"]                       # remove a key

for key, value in person.items():
    print(key, "->", value)

print(list(person.keys()))
print(list(person.values()))

# .setdefault() -> get a key, or set+return a default if it's missing
counts = {}
for word in ["a", "b", "a", "c", "b", "a"]:
    counts[word] = counts.get(word, 0) + 1   # classic "counting" pattern
print(counts)

# Merging dicts (Python 3.9+)
defaults = {"theme": "dark", "lang": "en"}
overrides = {"lang": "fr"}
merged = defaults | overrides
print(merged)
```

```python
# =========================================================
# COMPREHENSIONS - concise, Pythonic way to build lists/dicts/sets
# =========================================================

squares = [x**2 for x in range(10)]                     # list comprehension
evens = [x for x in range(20) if x % 2 == 0]             # with a filter condition
pairs = [(x, y) for x in range(3) for y in range(2)]     # nested loops

squared_dict = {x: x**2 for x in range(5)}               # dict comprehension
unique_lengths = {len(word) for word in ["hi", "bye", "ok", "no"]}  # set comprehension

print(squares)
print(evens)
print(pairs)
print(squared_dict)
print(unique_lengths)

# Generator expression - like a list comprehension but LAZY (computed on demand)
# Use () instead of [] -> memory-efficient for large/infinite sequences
gen = (x**2 for x in range(1000000))   # doesn't compute anything yet!
print(next(gen), next(gen))            # computes values one at a time
```
