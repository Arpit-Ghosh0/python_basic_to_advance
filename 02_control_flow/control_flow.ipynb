# 2. Control Flow

Controls the *order* in which code executes: conditionals (branching) and loops (repetition).
Python uses **indentation** (not braces `{}`) to define code blocks — this is not a style
choice, it's required syntax.


```python
# =========================================================
# IF / ELIF / ELSE
# =========================================================

age = 20

if age < 13:
    category = "child"
elif age < 20:
    category = "teenager"
else:
    category = "adult"

print(category)

# Ternary (conditional) expression - a compact one-line if/else
label = "adult" if age >= 18 else "minor"
print(label)

# Chained comparisons (Pythonic!)
x = 5
print(0 < x < 10)   # equivalent to: 0 < x and x < 10
```

```python
# =========================================================
# FOR LOOPS
# =========================================================

# Looping over a range of numbers
for i in range(5):          # 0,1,2,3,4  (stop is exclusive)
    print(i, end=" ")
print()

for i in range(2, 10, 2):   # start, stop, step -> 2,4,6,8
    print(i, end=" ")
print()

# Looping directly over a collection (the Pythonic way - no manual indexing needed)
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(fruit)

# enumerate() -> get both index AND value
for index, fruit in enumerate(fruits):
    print(index, fruit)

# zip() -> loop over multiple iterables in parallel
names = ["Alice", "Bob"]
scores = [90, 85]
for name, score in zip(names, scores):
    print(f"{name}: {score}")
```

```python
# =========================================================
# WHILE LOOPS
# =========================================================

count = 0
while count < 5:
    print(count, end=" ")
    count += 1
print()

# while True + break is a common "loop until condition met" pattern
count = 0
while True:
    if count >= 3:
        break        # exits the loop immediately
    print("looping", count)
    count += 1
```

```python
# =========================================================
# BREAK, CONTINUE, PASS
# =========================================================

for i in range(10):
    if i == 3:
        continue     # skip the rest of this iteration, go to next
    if i == 6:
        break        # exit the loop entirely
    print(i, end=" ")
print()

# pass = "do nothing" placeholder - useful for stub functions/classes while planning
def todo_function():
    pass   # will implement later; without this, empty body = SyntaxError
```

```python
# =========================================================
# MATCH-CASE (structural pattern matching, Python 3.10+)
# =========================================================
# Similar to switch/case in other languages, but more powerful.

def describe(value):
    match value:
        case 0:
            return "zero"
        case 1 | 2 | 3:                  # OR pattern
            return "small number"
        case int() if value > 100:        # guard clause
            return "big number"
        case [x, y]:                      # matches a 2-element list, unpacks it
            return f"a pair: {x} and {y}"
        case {"name": n}:                 # matches a dict with a 'name' key
            return f"a dict with name={n}"
        case _:                           # default / wildcard case
            return "something else"

for v in (0, 2, 500, [1, 2], {"name": "Eve"}, "hi"):
    print(v, "->", describe(v))
```
