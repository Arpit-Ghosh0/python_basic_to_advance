# 7. Object-Oriented Programming (OOP)

OOP models code around **objects** — bundles of data (attributes) and behavior (methods).
Python's four pillars: **Encapsulation, Inheritance, Polymorphism, Abstraction**.


```python
# =========================================================
# CLASSES & OBJECTS
# =========================================================

class Dog:
    species = "Canis familiaris"   # CLASS attribute - shared by all instances

    def __init__(self, name, age):
        # __init__ is the CONSTRUCTOR - runs automatically when you create an object
        # 'self' refers to the specific instance being created
        self.name = name           # INSTANCE attribute - unique to each object
        self.age = age

    def bark(self):                # INSTANCE method - needs 'self' to access instance data
        return f"{self.name} says Woof!"

    def __str__(self):
        # controls what print(object) / str(object) shows - a "human readable" representation
        return f"Dog(name={self.name}, age={self.age})"

d1 = Dog("Rex", 3)
d2 = Dog("Buddy", 5)

print(d1.bark())
print(d1)                 # uses __str__
print(d1.species, d2.species)   # both share the class attribute
```

```python
# =========================================================
# INHERITANCE - a class can reuse & extend another class
# =========================================================

class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        return f"{self.name} makes a sound"

class Cat(Animal):                  # Cat INHERITS from Animal
    def speak(self):                # method OVERRIDING - replaces the parent's version
        return f"{self.name} says Meow!"

class Puppy(Animal):
    def __init__(self, name, breed):
        super().__init__(name)      # calls the PARENT's __init__ to avoid duplicating logic
        self.breed = breed

    def speak(self):
        parent_sound = super().speak()   # can still call the parent's method if needed
        return f"{parent_sound} (a {self.breed} puppy)"

c = Cat("Whiskers")
p = Puppy("Max", "Labrador")
print(c.speak())
print(p.speak())

# POLYMORPHISM: same method name, different behavior per class
for animal in (c, p, Animal("Generic")):
    print(animal.speak())
```

```python
# =========================================================
# ENCAPSULATION - controlling access to internal data
# =========================================================

class BankAccount:
    def __init__(self, balance):
        self._balance = balance      # convention: single underscore = "protected" (internal use)
        self.__pin = "1234"          # double underscore = "name-mangled" (harder to access from outside)

    @property                        # turns a method into a READ-ONLY attribute-like access
    def balance(self):
        return self._balance

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Deposit amount must be positive")
        self._balance += amount

account = BankAccount(100)
print(account.balance)      # accessed like an attribute, but it's actually a method (via @property)
account.deposit(50)
print(account.balance)
# account.balance = 1000    # would raise AttributeError - no setter defined -> read-only
```

```python
# =========================================================
# CLASS METHODS, STATIC METHODS, PROPERTIES
# =========================================================

class Circle:
    def __init__(self, radius):
        self.radius = radius

    @property
    def area(self):
        # computed on the fly, always in sync with self.radius - looks like an attribute
        return 3.14159 * self.radius ** 2

    @classmethod
    def from_diameter(cls, diameter):
        # alternative constructor - 'cls' refers to the class itself, not an instance
        return cls(diameter / 2)

    @staticmethod
    def is_valid_radius(value):
        # doesn't need 'self' or 'cls' - just a utility function grouped inside the class
        return value > 0

c = Circle(5)
print(c.area)

c2 = Circle.from_diameter(10)   # created via the alternative constructor
print(c2.radius)

print(Circle.is_valid_radius(-5))
```

```python
# =========================================================
# MAGIC / DUNDER METHODS - let your objects work with built-in syntax
# =========================================================

class Vector:
    def __init__(self, x, y):
        self.x, self.y = x, y

    def __add__(self, other):        # enables: vector1 + vector2
        return Vector(self.x + other.x, self.y + other.y)

    def __eq__(self, other):         # enables: vector1 == vector2
        return self.x == other.x and self.y == other.y

    def __repr__(self):              # unambiguous representation, shown in REPL/debugger
        return f"Vector({self.x}, {self.y})"

    def __len__(self):               # enables: len(vector)
        return int((self.x ** 2 + self.y ** 2) ** 0.5)

v1 = Vector(1, 2)
v2 = Vector(3, 4)
print(v1 + v2)          # Vector(4, 6)   -> via __add__
print(v1 == Vector(1, 2))   # True       -> via __eq__
print(len(v2))           # 5 (magnitude, via __len__)
```

```python
# =========================================================
# ABSTRACT BASE CLASSES - enforce that subclasses implement certain methods
# =========================================================
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass   # no implementation here - subclasses MUST provide one

class Square(Shape):
    def __init__(self, side):
        self.side = side

    def area(self):
        return self.side ** 2

# shape = Shape()      # would raise TypeError - can't instantiate an abstract class
sq = Square(4)
print(sq.area())
```
