# 5. Modules & Packages

A **module** is just a `.py` file. A **package** is a folder of modules with an `__init__.py`
(implicit in modern Python). This is how you organize and reuse code across files.


````python
# =========================================================
# IMPORTING MODULES
# =========================================================

import math
print(math.sqrt(16), math.pi)

from math import sqrt, pi          # import specific names directly (no "math." prefix needed)
print(sqrt(25))

import math as m                   # alias, common for long/frequent names (e.g. numpy as np)
print(m.floor(4.7))

from math import *                 # imports EVERYTHING - generally discouraged
                                    # (pollutes the namespace, unclear where names came from)

# Some commonly used standard library modules worth knowing:
import random
import datetime
import os
import sys
import json
import collections

print(random.randint(1, 10))            # random int between 1 and 10 (inclusive)
print(datetime.date.today())            # today's date
print(os.getcwd())                      # current working directory
print(sys.version.split()[0])           # Python version
print(collections.Counter("mississippi"))   # counts occurrences of each character
````

````python
# =========================================================
# THE __name__ == "__main__" PATTERN
# =========================================================
# Every module has a built-in __name__ variable.
# - If the file is run DIRECTLY, __name__ == "__main__"
# - If the file is IMPORTED by another file, __name__ == "<module name>"
# This lets a file be both a reusable module AND a runnable script.

print(__name__)   # in a notebook this prints '__main__'

# Typical structure in a real .py file:
"""
def main():
    print("Running as the main script")

if __name__ == "__main__":
    main()   # only runs when this file is executed directly, NOT when imported
"""
````
