# 6. File Handling

Reading and writing files. Always prefer the `with` statement — it automatically closes the
file for you, even if an error occurs.

```python
# =========================================================
# WRITING & READING TEXT FILES
# =========================================================

# "w" = write (overwrites the file if it exists), "a" = append, "r" = read (default)
with open("sample.txt", "w") as f:
    f.write("Hello, file!\n")
    f.write("Second line.\n")
# file is automatically closed here, even if an exception happened above

with open("sample.txt", "r") as f:
    content = f.read()          # read the WHOLE file as one string
print(content)

with open("sample.txt", "r") as f:
    for line in f:               # memory-efficient: reads line by line
        print(line.strip())      # .strip() removes the trailing newline

with open("sample.txt", "a") as f:
    f.write("Appended line.\n")

with open("sample.txt", "r") as f:
    lines = f.readlines()        # read all lines into a list of strings
print(lines)
```

```python
# =========================================================
# WORKING WITH JSON (very common for config files / APIs)
# =========================================================
import json

data = {"name": "Alice", "age": 30, "hobbies": ["reading", "coding"]}

with open("data.json", "w") as f:
    json.dump(data, f, indent=2)     # write a dict to a JSON file (indent = pretty-print)

with open("data.json", "r") as f:
    loaded = json.load(f)            # read JSON back into a Python dict
print(loaded)

# Convert to/from a JSON *string* (not a file) with dumps/loads
json_string = json.dumps(data)
print(json_string)
print(json.loads(json_string))
```

```python
# =========================================================
# WORKING WITH CSV
# =========================================================
import csv

rows = [["name", "age"], ["Alice", 30], ["Bob", 25]]

with open("people.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerows(rows)

with open("people.csv", "r") as f:
    reader = csv.reader(f)
    for row in reader:
        print(row)

# DictReader/DictWriter map rows to/from dicts using the header row as keys
with open("people.csv", "r") as f:
    for row in csv.DictReader(f):
        print(row)   # {'name': 'Alice', 'age': '30'}
```
