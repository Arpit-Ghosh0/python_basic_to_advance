# 8. Exception Handling

Exceptions let you gracefully handle errors instead of crashing the whole program.

```python
# =========================================================
# TRY / EXCEPT / ELSE / FINALLY
# =========================================================

def divide(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        print("Error: cannot divide by zero!")
        return None
    except TypeError as e:
        print(f"Error: invalid types - {e}")
        return None
    else:
        # runs ONLY if the try block succeeded with no exception
        print("Division succeeded")
        return result
    finally:
        # ALWAYS runs, whether an exception happened or not - great for cleanup
        print("Divide operation attempted.")

print(divide(10, 2))
print(divide(10, 0))
print(divide(10, "a"))
```

```python
# =========================================================
# CATCHING MULTIPLE / GENERIC EXCEPTIONS, AND RAISING YOUR OWN
# =========================================================

try:
    numbers = [1, 2, 3]
    print(numbers[10])
except (IndexError, KeyError) as e:      # catch several exception types at once
    print(f"Lookup error: {e}")
except Exception as e:
    # generic catch-all - use sparingly, prefer catching SPECIFIC exceptions
    print(f"Unexpected error: {e}")

def set_age(age):
    if age < 0:
        raise ValueError("Age cannot be negative")   # manually raise an exception
    return age

try:
    set_age(-5)
except ValueError as e:
    print(f"Caught: {e}")
```

```python
# =========================================================
# CUSTOM EXCEPTIONS - define your own error types for clearer error handling
# =========================================================

class InsufficientFundsError(Exception):
    """Raised when a withdrawal exceeds the available balance."""
    def __init__(self, balance, amount):
        self.balance = balance
        self.amount = amount
        super().__init__(f"Cannot withdraw {amount}, only {balance} available")

def withdraw(balance, amount):
    if amount > balance:
        raise InsufficientFundsError(balance, amount)
    return balance - amount

try:
    withdraw(100, 150)
except InsufficientFundsError as e:
    print(f"Transaction failed: {e}")
```
